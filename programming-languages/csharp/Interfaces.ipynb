Her ser vi hvordan man kan bytte servicen ud, uden at ændre på koden der bruger den.

```polyglot-notebook
// KONTRAKTEN: Hvad skal man kunne?
public interface IGreeter
{
    void SayHello();
}

// RIGTIG implementation
public class FriendlyGreeter : IGreeter
{
    public void SayHello() => Console.WriteLine("Hej med dig, min ven! 👋");
}

// TEST implementation (Mock/Stub)
public class AngryGreeter : IGreeter
{
    public void SayHello() => Console.WriteLine("GÅ VÆK! 😠");
}

// En funktion der er ligeglad med hvem den får, bare den overholder kontrakten
void RunApp(IGreeter greeter)
{
    Console.Write("Programmet siger: ");
    greeter.SayHello();
}

// Vi kan skifte adfærd nemt:
RunApp(new FriendlyGreeter());
RunApp(new AngryGreeter());
```
