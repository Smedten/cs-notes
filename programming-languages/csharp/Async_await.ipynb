Vi brugte async Task og await i App.cs for at få spinneren til at køre glidende.

Teori: I en CLI (ligesom i web apps) vil vi undgå at blokere den tråd, der tegner brugergrænsefladen. Hvis vi brugte Thread.Sleep(1000), ville hele terminalen fryse. Når vi bruger await Task.Delay(1000), frigives tråden til at lave andre ting (som at tegne spinner-animationen), mens den venter på, at tiden går.

```polyglot-notebook
// Top-level statements (kører med det samme)
await Task.Delay(1000); 

// Metode definition (gemmes til senere brug)
// Bemærk: I notebooks behøver du ikke "public"
async Task GoryMyWorkAsync()
{
    Console.WriteLine("Starter arbejde...");
    await Task.Delay(2000); 
    Console.WriteLine("Færdig!");
}

await GoryMyWorkAsync();
```

```polyglot-notebook
// Simulering af en langsom kok (Task)
async Task<string> LavBurgerAsync()
{
    Console.WriteLine("Kok: Jeg går i gang...");
    await Task.Delay(2000); // Venter 2 sekunder (men blokerer ikke)
    return "🍔 Her er din burger!";
}

Console.WriteLine("Mig: Jeg bestiller...");
Task<string> bipper = LavBurgerAsync(); // Jeg får bipperen (Tasken)

Console.WriteLine("Mig: Jeg tjekker Instagram mens jeg venter...");
// Her kan programmet lave andet arbejde

string mad = await bipper; // Nu bipper den, og jeg henter maden
Console.WriteLine(mad);
```
