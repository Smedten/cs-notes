### Stack vs. Heap (Hukommelsen)

The Stack (Stakken): En meget lille, lynhurtig hukommelse. Her bor simple værdier (tal, booleans) og variabler. Tænk på det som Post-it noter på dit skrivebord.

The Heap (Dungen): En kæmpe lagerhal til store objekter. Her bor Classes, Strings og lister.

### Value Types (Værdityper):

Eksempler: int, bool, double, struct.

Bor direkte på Stacken.

Når du kopierer (a = b), tager du en fotokopi af værdien. De er helt uafhængige.

### Reference Types (Referencetyper):

Eksempler: class, interface, delegate, string, list.

Selve dataen bor ude i lagerhallen (Heap).

På dit skrivebord (Stack) har du kun en adresse (en pointer/fjernbetjening) til lagerhallen.

Når du kopierer (a = b), kopierer du kun fjernbetjeningen. Begge variabler peger nu på det samme objekt i lagerhallen!

### Records - Den magiske hybrid:

En record er teknisk set en Reference Type (den bor på Heapen).

MEN: Compileren snyder og får den til at opføre sig som en Value Type. Når du sammenligner dem, tjekker den indholdet, ikke adressen. Når du bruger with, laver den en ny kopi i stedet for at ændre den gamle.

```polyglot-notebook
// int er en Value Type
int a = 10;
int b = a; // Vi tager en fotokopi af værdien 10

b = 20; // Vi ændrer kun kopien

Console.WriteLine($"a er: {a}"); // a er stadig 10!
Console.WriteLine($"b er: {b}");
```

Reference Types (Kopier er farlige!) Dette er her, mange fejler.

```polyglot-notebook
public class PersonClass 
{ 
    public string Navn { get; set; } 
}

// Vi opretter et objekt på HEAPEN. p1 er bare en fjernbetjening.
var p1 = new PersonClass { Navn = "Anders" };

// Vi kopierer fjernbetjeningen! Nu har vi to fjernbetjeninger til SAMME TV.
var p2 = p1; 

// Vi bruger fjernbetjening 2 til at skifte kanal (ændre navn)
p2.Navn = "Bente";

// Hvad hedder p1 nu?
Console.WriteLine($"p1 hedder: {p1.Navn}"); // CHOK! p1 er også Bente nu!
Console.WriteLine($"p2 hedder: {p2.Navn}");
```

Records (Best of both worlds) Her ser du, hvorfor records er geniale til data. De er Reference Types (så de er hurtige at sende rundt), men de er Immutable (uforanderlige), så du kan ikke lave "Bente-fejlen" ovenover.

```polyglot-notebook
public record PersonRecord(string Navn);

var r1 = new PersonRecord("Anders");
var r2 = r1; // Kopierer referencen (peger på samme sted)

// r2.Navn = "Bente"; // FEJL! Compileren forbyder dig at ændre den.

// Du er tvunget til at lave et NYT objekt, hvis du vil ændre noget.
var r3 = r1 with { Navn = "Bente" };

Console.WriteLine($"r1 er: {r1.Navn}"); // Anders (Sikker!)
Console.WriteLine($"r3 er: {r3.Navn}"); // Bente (Ny kopi)
```
