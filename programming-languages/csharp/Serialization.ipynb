```polyglot-notebook
using System.Text.Json; // Det moderne bibliotek i .NET 8

// Vores C# objekt
public record Opgave(int Id, string Navn);
var minOpgave = new Opgave(1, "Køb Kaffe");

// 1. SERIALIZATION (C# -> JSON String)
// Vi laver objektet om til tekst, der kan gemmes eller sendes over nettet
string jsonText = JsonSerializer.Serialize(minOpgave);
Console.WriteLine($"JSON Output: {jsonText}");
// Output: {"Id":1,"Navn":"Køb Kaffe"}
 

// 2. DESERIALIZATION (JSON String -> C#)
// Vi læser tekst og laver det om til et levende objekt igen
var nyOpgave = JsonSerializer.Deserialize<Opgave>(jsonText);
Console.WriteLine($"C# Objekt navn: {nyOpgave.Navn}");
```
