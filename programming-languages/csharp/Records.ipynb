Records er perfekt til at gemme data fordi de indeholder rå data frem for hukommelses-location (som klasser).

```polyglot-notebook
// Definition af en record (Data container)
public record Person(string Name, int Age);

// Oprettelse
var p1 = new Person("Hans", 30);
var p2 = new Person("Hans", 30); // Identiske data

// Sammenligning
// Med classes ville dette være False (fordi det er to forskellige objekter i ram)
// Med records er det True (fordi dataen er ens)
Console.WriteLine($"Er p1 og p2 ens? {p1 == p2}"); 

// "Nondestructive mutation" (with)
// Laver en kopi, ændrer én ting, og lader den gamle være
var p3 = p1 with { Age = 31 };
Console.WriteLine($"P1 alder: {p1.Age}, P3 alder: {p3.Age}");
```

```polyglot-notebook
// CLASS (Ler - ændrer sig)
public class ClassOpgave { public string Navn { get; set; } }

var c1 = new ClassOpgave { Navn = "Vask Op" };
var c2 = c1; // c2 peger på SAMME lerklump
c2.Navn = "Slå Græs"; 

// Hvad hedder c1 nu?
Console.WriteLine($"Class c1 hedder nu: {c1.Navn}"); // Den blev ændret "bag ryggen" på c1!


// RECORD (Kvittering - fastlåst)
public record RecordOpgave(string Navn);

var r1 = new RecordOpgave("Vask Op");
// r1.Navn = "Slå Græs"; // FEJL! Du må ikke rette i den.

// Vi laver en NY kopi med ændringen (Non-destructive mutation)
var r2 = r1 with { Navn = "Slå Græs" };

Console.WriteLine($"Record r1 er stadig: {r1.Navn}"); // Uændret!
Console.WriteLine($"Record r2 er ny: {r2.Navn}");
```
