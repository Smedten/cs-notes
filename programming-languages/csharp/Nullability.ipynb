Hvordan vi sikre os mod fejl, når dataen mangler (er null).
Vi pakker den data som måske er null i en lille wrapper, som siger enten er vi "data" eller også er vi null.

```polyglot-notebook
// DateTime? betyder "Kan være en dato ELLER null"
DateTime? deadline = null;

if (deadline.HasValue)
{
    Console.WriteLine($"Deadline er: {deadline.Value}");
}
else
{
    Console.WriteLine("Ingen deadline sat.");
}

// Null-coalescing operator (??)
// Betyder: "Brug deadline, men hvis den er null, så brug DateTime.Now"
DateTime finalDate = deadline ?? DateTime.Now;
Console.WriteLine($"Vi bruger datoen: {finalDate}");
```
